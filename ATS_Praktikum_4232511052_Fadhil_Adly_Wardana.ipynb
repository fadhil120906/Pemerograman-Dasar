```python
# SOAL 1: Daya Generator dan Efisiensi Turbin

tegangan = 220       # Volt
arus = 15            # Ampere
energi_input = 500   # Watt
energi_output = 400  # Watt

# a
daya_listrik = tegangan * arus
print(f"Daya Listrik: {daya_listrik} W")

# b.
efisiensi = (energi_output / energi_input) * 100
print(f"Efisiensi: {efisiensi:.2f}%")
print("-" * 50)
```

```python
# SOAL 5: Katup Pengaman

tekanan_uap = float(input("Masukkan tekanan uap (psi): "))
status_manual_override = False  # Ubah menjadi True untuk menguji mode manual

if status_manual_override == True:
    print("Katup Mode Manual")
elif tekanan_uap > 150 and status_manual_override == False:
    print("Katup Terbuka")
else:
    print("Katup Tertutup")
print("-" * 50)
```

```python
# SOAL 7: Pemanas Mencapai Target

suhu_saat_ini = 60

while suhu_saat_ini < 85:
    suhu_saat_ini += 5
    print(f"Suhu saat ini: {suhu_saat_ini} C")

    if suhu_saat_ini == 80:
        print("WARNING: Mendekati target!")

print("Target 85 C tercapai.")
print("-" * 50)
```

```python
# SOAL 9: Pembacaan Vibrasi dengan Sensor Rusak

data = ["1.8", "2.4", "ERR", "3.2", "1.9"]
data_tidak_valid = 0

for i, nilai_str in enumerate(data, start=1):
    try:
        nilai = float(nilai_str)
    except ValueError:
        print(f"Unit {i}: data tidak valid")
        data_tidak_valid += 1
        continue

    if nilai > 2.0:
        print(f"Warning: Unit {i} vibrasi tinggi!")
        if nilai > 3.0:
            print("KRITIS: pemeriksaan dihentikan")
            break

print(f"Jumlah data tidak valid: {data_tidak_valid}")
print("-" * 50)
```
